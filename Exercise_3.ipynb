# Exercise 3 – News Article and Top Five Words



## Question
Read a `.txt` news article and identify the **five most visually prominent words**.

## Input File
This notebook creates and reads `news_article.txt` in the same folder. The article is about current uses of Artificial Intelligence in organizations, education, and healthcare.

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud, STOPWORDS

filename = "news_article.txt"

with open(filename, "r", encoding="utf-8") as file:
    news_text = file.read()

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white",
    stopwords=set(STOPWORDS),
    min_font_size=10,
    random_state=42
).generate(news_text)

plt.figure(figsize=(15, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

```python
# Get the five most frequent words used by the WordCloud
frequencies = wordcloud.words_
top_five = list(frequencies.items())[:5]

print("Five most visually prominent words:")
for word, score in top_five:
    print(f"{word}: {score:.3f}")
```

## Answer
The five most visually prominent words are obtained from `wordcloud.words_`. The exact order is frequency-based, so the largest words in the generated cloud are the most prominent.
