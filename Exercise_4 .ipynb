# Exercise 4 – Word Cloud Parameter Experiment



## Question
Experiment with `max_words`, `min_font_size`, `width`, and `height`.

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud, STOPWORDS

text = """
Artificial Intelligence machine learning data analysis natural language processing
computer vision automation robotics chatbots education healthcare finance security
technology innovation algorithms models training prediction classification data
science software systems intelligent services decision support automation.
Artificial Intelligence and machine learning help organizations analyze data,
automate tasks and build intelligent applications for users.
"""

settings = [
    {"max_words": 50, "min_font_size": 10, "width": 800, "height": 400},
    {"max_words": 100, "min_font_size": 5, "width": 1000, "height": 500},
    {"max_words": 30, "min_font_size": 15, "width": 1200, "height": 600},
]

for i, s in enumerate(settings, 1):
    wc = WordCloud(
        background_color="white",
        stopwords=set(STOPWORDS),
        random_state=42,
        **s
    ).generate(text)

    plt.figure(figsize=(12, 6))
    plt.imshow(wc, interpolation="bilinear")
    plt.title(f"Experiment {i}: {s}")
    plt.axis("off")
    plt.show()
```

## Explanation

- **`max_words`** controls the maximum number of words displayed. A smaller value gives a simpler cloud.
- **`min_font_size`** controls the smallest font size allowed. A larger value can remove very small words.
- **`width`** controls the image width.
- **`height`** controls the image height.

Therefore, changing these parameters changes the number, size, and shape of words in the Word Cloud.
