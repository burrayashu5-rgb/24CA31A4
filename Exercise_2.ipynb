# Exercise 2 – Customer Reviews with Stopword Removal



## Question
Create a Word Cloud from **10 customer reviews** and remove English stopwords.

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud, STOPWORDS

reviews = [
    "The product quality is excellent and the delivery was fast.",
    "I am happy with the product and the service was excellent.",
    "The delivery was quick and the packaging was good.",
    "The product quality is good and works perfectly.",
    "Customer service was helpful and friendly.",
    "I received the product on time and the quality is excellent.",
    "The product is easy to use and very useful.",
    "Good quality, fast delivery and excellent service.",
    "I am satisfied with the product and customer support.",
    "The packaging was safe and the product quality was great."
]

text = " ".join(reviews)

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white",
    stopwords=set(STOPWORDS),
    min_font_size=10,
    random_state=42
).generate(text)

plt.figure(figsize=(15, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

## Answer / Observation
English stopwords such as **the, was, and, I, with** are removed. Meaningful review words such as **product, quality, delivery, excellent, service, and good** become easier to see.
