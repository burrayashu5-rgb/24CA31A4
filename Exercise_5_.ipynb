# Exercise 5 – Before and After Stopword Removal



## Question
Compare the Word Cloud before and after stopword removal and explain the difference.

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud, STOPWORDS

text = """
Artificial Intelligence is a powerful technology and it is used in many areas.
The technology helps people and organizations to analyze data and improve
services. AI is used in chatbots, education, healthcare, finance and automation.
It can help computers understand language and images. The use of AI is growing
and the technology is becoming important for modern applications.
"""

without_stopwords = WordCloud(
    width=900,
    height=450,
    background_color="white",
    random_state=42
).generate(text)

with_stopwords_removed = WordCloud(
    width=900,
    height=450,
    background_color="white",
    stopwords=set(STOPWORDS),
    random_state=42
).generate(text)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].imshow(without_stopwords, interpolation="bilinear")
axes[0].set_title("Before Stopword Removal")
axes[0].axis("off")

axes[1].imshow(with_stopwords_removed, interpolation="bilinear")
axes[1].set_title("After Stopword Removal")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

## Comparison / Answer

**Before stopword removal:** Common English words such as **the, is, and, it, in, to** can appear frequently and may occupy noticeable space.

**After stopword removal:** These common words are excluded, so meaningful content words such as **Artificial, Intelligence, technology, data, AI, chatbots, healthcare, education,** and **automation** become easier to identify.

**Conclusion:** Stopword removal makes the Word Cloud clearer and more useful for quick analysis because it gives more visual attention to meaningful words.
